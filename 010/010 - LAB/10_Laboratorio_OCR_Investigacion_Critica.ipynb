## Terminología clave (Microglosario)

* **OCR (Reconocimiento Óptico de Caracteres):** Técnica que permite extraer texto a partir de imágenes, escaneos o fotografías de documentos.
* **Fuente primaria:** Recurso técnico original donde una afirmación puede verificarse directamente, por ejemplo la documentación oficial, el repositorio de un modelo o un paper.
* **Handle de modelo:** Identificador exacto con el que un modelo se publica en plataformas como ✦ 🤗 Hugging Face.
* **Pipeline:** Abstracción de `transformers` que simplifica la carga del modelo, el preprocesamiento y la inferencia.
* **CER / WER:** Métricas de error para comparar el texto reconocido con un texto de referencia. CER mide error por carácter y WER por palabra.
* **Trazabilidad:** Registro explícito de qué consultaron, qué fuente usaron, qué verificaron y qué decisión tomaron.

Mientras verifican, deténganse en estas preguntas:

1. ¿Todas las afirmaciones equivocadas son igual de graves para un desarrollo real?
2. ¿Qué diferencia encontraron entre un dato incorrecto, uno impreciso y uno no verificable?
3. Si un handle está mal escrito pero la idea general parece razonable, ¿alcanza con eso para implementarlo? ¿Por qué?

---

**Respuestas:**

**1.** No todas tienen el mismo impacto. Un handle incorrecto (como `baidu-nlp/clova-ocr-vietnamese`) hace que el código falle al instante con un error de "model not found" — es un error de alta visibilidad, fácil de detectar. En cambio, un error arquitectónico como "decoder = GPT-2" cuando en realidad es RoBERTa no rompe el pipeline, pero puede llevar a decisiones incorrectas de fine-tuning o a diseñar mal un postprocesador de salida esperando tokenización de GPT-2. El error de sintaxis `result['generated_text']` cae a mitad de camino: falla en ejecución pero solo si no se testa el output.

**2.** 
- **Incorrecto:** la afirmación es factualmente falsa y verificable como tal (ej: atribuir CLOVA a Baidu, o decir GPT-2 cuando es RoBERTa).
- **Impreciso:** la idea de fondo es correcta pero el detalle está mal expresado (ej: llamar "DocVQA" al tag cuando DocVQA es un benchmark separado).
- **No verificable:** no hay fuente primaria que lo confirme ni lo refute con certeza — en este caso ninguna afirmación cayó en esa categoría, pero ocurre con claims sobre "número de parámetros" de versiones informales o modelos privados.

**3.** No alcanza. Un handle incorrecto hace que `pipeline()` lance `OSError` o `RepositoryNotFoundError` sin ejecutar nada. Además, si el handle está mal puede estar apuntando a un modelo completamente diferente que casualmente existe con ese nombre, generando resultados silenciosamente incorrectos. La verificación directa en la página del modelo tarda menos de 30 segundos y evita horas de debugging.

### Preparación de imágenes de prueba

Para esta actividad conviene trabajar con dos imágenes contrastantes. Pueden usar archivos propios o descargar muestras desde una fuente confiable. La recomendación es registrar siempre de dónde salió cada imagen.

Sugerencia de organización:

* `imagen_impresa.png`
* `imagen_mano.png`

Si no tienen imágenes propias, creen una carpeta local llamada `ocr_muestras/` y guarden allí las dos muestras que vayan a usar.

````python
from pathlib import Path
from PIL import Image, ImageDraw, ImageFont, ImageFilter
import numpy as np

carpeta_muestras = Path("ocr_muestras")
carpeta_muestras.mkdir(exist_ok=True)

ruta_imagen_impresa = carpeta_muestras / "imagen_impresa.png"
ruta_imagen_mano    = carpeta_muestras / "imagen_mano.png"

# ── Imagen 1: texto impreso limpio ──────────────────────────────────────────
if not ruta_imagen_impresa.exists():
    img_impresa = Image.new("RGB", (512, 64), color=(255, 255, 255))
    draw = ImageDraw.Draw(img_impresa)
    # PIL incluye una fuente bitmap por defecto
    draw.text((10, 15), "Processing digital images with Python", fill=(0, 0, 0))
    img_impresa.save(ruta_imagen_impresa)
    print(f"✓ Imagen impresa creada: {ruta_imagen_impresa}")
else:
    print(f"✓ Imagen impresa ya existe: {ruta_imagen_impresa}")

# ── Imagen 2: texto "degradado" (ruido + blur para simular manuscrito/escáner) ─
if not ruta_imagen_mano.exists():
    img_base = Image.new("RGB", (512, 64), color=(245, 242, 235))
    draw = ImageDraw.Draw(img_base)
    draw.text((10, 15), "Handwritten style with noise", fill=(30, 30, 60))
    # Aplicar ruido gaussiano para simular degradación
    arr = np.array(img_base, dtype=np.float32)
    ruido = np.random.normal(0, 18, arr.shape)
    arr = np.clip(arr + ruido, 0, 255).astype(np.uint8)
    img_degradada = Image.fromarray(arr).filter(ImageFilter.GaussianBlur(radius=0.8))
    img_degradada.save(ruta_imagen_mano)
    print(f"✓ Imagen degradada creada: {ruta_imagen_mano}")
else:
    print(f"✓ Imagen degradada ya existe: {ruta_imagen_mano}")

print("\nNota: estas imágenes fueron generadas programáticamente para la prueba.")
print("Para un test más realista, reemplazarlas con fotos de documentos reales.")
````

````python
print("✦ Abriendo imágenes de prueba...")

imagen_impresa = Image.open(ruta_imagen_impresa)
imagen_mano    = Image.open(ruta_imagen_mano)

print(f"✓ imagen_impresa: {imagen_impresa.size} px, modo {imagen_impresa.mode}")
print(f"✓ imagen_mano:    {imagen_mano.size} px, modo {imagen_mano.mode}")
````

**Respuestas:**

**1.** Las fuentes más confiables resultaron ser: (a) la **página del modelo en Hugging Face** (nombre de la organización, handle exacto, fecha de última actualización), y (b) el **paper original** para información arquitectónica. Las fuentes secundarias como tutoriales y posts de Medium son útiles para entender el flujo general pero no para verificar detalles exactos de nombres o parámetros.

**2.** El criterio principal fue la **trazabilidad directa**: si podía ir a una URL y leer el dato exacto, la afirmación era verificable. Para distinguir incorrecto de impreciso usé la pregunta: "¿la idea de fondo es correcta aunque el detalle falle?". Si la idea central es falsa (atribuir CLOVA a Baidu), es incorrecto. **4.** La próxima vez empezaría por abrir el repositorio o la documentación oficial **antes** de escribir código. Un minuto de verificación del handle en Hugging Face evita diez minutos de debugging de errores crípticos de carga de modelo.

**5.** La transferencia más importante es el **protocolo de auditoría**: extraer afirmaciones verificables → identificar fuente primaria → buscar evidencia directa → clasificar. Este flujo aplica igual si el tema es una librería nueva, una API de terceros o un modelo de ML que nunca usé. La habilidad no es saber OCR; es saber cómo verificar cualquier claim técnico antes de construir sobre él.
