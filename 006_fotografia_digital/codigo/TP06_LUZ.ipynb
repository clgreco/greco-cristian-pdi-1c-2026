```python
from pathlib import Path
import cv2
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.patches as patches
from matplotlib.patches import FancyArrowPatch
```

# Punto 7 — Fotografía basada en la luz

**Objetivo:** Usar la luz como elemento estructural de la imagen.

Las fotografías fueron tomadas durante la **hora dorada de la mañana** desde un balcón en Buenos Aires. La luz ingresa de forma lateral-frontal a baja altura, cálida y suave, envolviendo las plantas con tonos dorados y generando sombras largas y difusas que construyen volumen y atmósfera.

```python
# Rutas de las imágenes

ruta_img3  = Path("punto07_IMG_3.jpeg")
ruta_img20 = Path("punto07_IMG_20.jpeg")


def cargar_rgb(ruta):
    """
    Carga una imagen y la convierte de BGR a RGB
    para visualizarla correctamente con Matplotlib.
    """
    imagen_bgr = cv2.imread(
        str(ruta),
        cv2.IMREAD_COLOR
    )

    if imagen_bgr is None:
        raise FileNotFoundError(
            f"No se pudo leer la imagen: {ruta}"
        )

    imagen_rgb = cv2.cvtColor(
        imagen_bgr,
        cv2.COLOR_BGR2RGB
    )

    return imagen_rgb


imagen_a = cargar_rgb(ruta_img3)
imagen_b = cargar_rgb(ruta_img20)

print(f"Imagen A shape: {imagen_a.shape}")
print(f"Imagen B shape: {imagen_b.shape}")
```

```python
# =========================
# 1. VISUALIZACIÓN DE LAS FOTOGRAFÍAS ORIGINALES
# =========================

fig, axes = plt.subplots(1, 2, figsize=(16, 9))

axes[0].imshow(imagen_a)
axes[0].set_title(
    "Fotografía A — Hora dorada (encuadre cercano)",
    fontweight="bold"
)
axes[0].axis("off")

axes[1].imshow(imagen_b)
axes[1].set_title(
    "Fotografía B — Hora dorada (encuadre amplio)",
    fontweight="bold"
)
axes[1].axis("off")

plt.suptitle(
    "Fotografías basadas en la luz — Hora dorada de la mañana",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

Las fotografías fueron tomadas durante la **hora dorada de la mañana** (golden hour), momento en que el sol está bajo sobre el horizonte y la luz recorre una mayor distancia atmosférica antes de llegar al sujeto.

Esto produce:
- **Luz cálida:** tonos anaranjados y dorados que se perciben sobre el cielo y los bordes de las hojas.
- **Luz lateral-baja:** el sol ingresa desde el horizonte, iluminando lateralmente los sujetos y generando sombras largas y suaves.
- **Bajo contraste relativo:** la atmósfera dispersa la luz, suavizando las sombras.
- **Atmósfera:** el cielo degradado de naranja a azul construye un fondo con profundidad emocional.

La imagen elegida como principal es la **Fotografía A** (encuadre cercano), donde la maceta grande en primer plano concentra la atención y el cielo dorado actúa como fondo narrativo.

```python
# =========================
# 2. HISTOGRAMA RGB — IMAGEN PRINCIPAL (A)
# =========================

plt.figure(figsize=(8, 4))

colores = ("red", "green", "blue")
nombres = ("Rojo", "Verde", "Azul")

for i, (color, nombre) in enumerate(zip(colores, nombres)):

    histograma = cv2.calcHist(
        [imagen_a],
        [i],
        None,
        [256],
        [0, 256]
    )

    plt.plot(
        histograma,
        color=color,
        label=nombre
    )

plt.title(
    "Histograma RGB — Fotografía A (hora dorada)",
    fontweight="bold"
)

plt.xlabel("Intensidad")
plt.ylabel("Cantidad de píxeles")
plt.legend()
plt.tight_layout()
plt.show()

print(
    "Nota: el canal Rojo desplazado hacia valores altos confirma"
    " la dominante cálida característica de la hora dorada."
)
```

El histograma RGB revela la **dominante cálida** de la hora dorada:
el canal Rojo concentra más píxeles en valores altos (iluminación cálida),
mientras que el canal Azul se desplaza hacia valores medios-bajos,
confirmando que la escena está bañada por luz naranja-dorada.

```python
# =========================
# 3. CONVERSIÓN A ESCALA DE GRISES
# =========================

imagen_a_gris = cv2.cvtColor(
    imagen_a,
    cv2.COLOR_RGB2GRAY
)


fig, axes = plt.subplots(1, 2, figsize=(14, 7))

axes[0].imshow(imagen_a)
axes[0].set_title(
    "Original (color)",
    fontweight="bold"
)
axes[0].axis("off")

axes[1].imshow(imagen_a_gris, cmap="gray")
axes[1].set_title(
    "Escala de grises",
    fontweight="bold"
)
axes[1].axis("off")

plt.suptitle(
    "Comparación: color vs. escala de grises",
    fontsize=13,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

En escala de grises desaparece la dominante cálida del atardecer, pero se hace **más evidente el contraste tonal** entre las zonas iluminadas (bordes de las hojas, cielo claro) y las zonas en sombra (interior de las macetas, suelo del balcón). Esto permite leer mejor el **volumen y la textura** de las hojas de la sansevieria.

```python
# =========================
# 4. HISTOGRAMA ESCALA DE GRISES
# =========================

plt.figure(figsize=(8, 4))

plt.hist(
    imagen_a_gris.ravel(),
    bins=256,
    range=(0, 256),
    color="steelblue",
    alpha=0.8
)

plt.title(
    "Histograma — Escala de grises",
    fontweight="bold"
)

plt.xlabel("Intensidad")
plt.ylabel("Cantidad de píxeles")
plt.tight_layout()
plt.show()
```

```python
# =========================
# 5. ESQUEMA DE DIRECCIÓN DE LUZ
# =========================

# Creamos una copia para dibujar el esquema
imagen_esquema = imagen_a.copy()

h, w = imagen_esquema.shape[:2]

# Flecha indicando dirección de la luz (desde horizonte lateral-bajo)
# La luz viene desde atrás-lateral-izquierda (horizonte)
origen = (int(w * 0.05), int(h * 0.55))
destino = (int(w * 0.35), int(h * 0.45))

cv2.arrowedLine(
    imagen_esquema,
    origen,
    destino,
    (255, 220, 50),   # amarillo dorado
    8,
    tipLength=0.12
)

# Etiqueta
cv2.putText(
    imagen_esquema,
    "Luz dorada (SE, bajo)",
    (int(w * 0.03), int(h * 0.50)),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.7,
    (255, 220, 50),
    2
)

# Segunda anotación: zona iluminada
cv2.putText(
    imagen_esquema,
    "Bordes iluminados",
    (int(w * 0.45), int(h * 0.22)),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.6,
    (255, 255, 255),
    2
)

# Sombras
cv2.putText(
    imagen_esquema,
    "Sombras interiores",
    (int(w * 0.05), int(h * 0.85)),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.6,
    (200, 200, 200),
    2
)

plt.figure(figsize=(7, 10))
plt.imshow(imagen_esquema)
plt.title(
    "Esquema de dirección e impacto de la luz",
    fontweight="bold"
)
plt.axis("off")
plt.tight_layout()
plt.show()
```

La flecha indica que la luz solar proviene desde el horizonte en dirección sureste, con una altura angular muy baja (característica de la hora dorada). Esto produce:

- **Bordes iluminados:** los extremos de las hojas de la sansevieria reciben luz directa y brillan con tonos cálidos.
- **Sombras interiores largas:** el interior de la maceta y el suelo quedan en sombra suave.
- **Cielo naranja degradado:** el fondo actúa como fuente secundaria de luz difusa.

```python
# =========================
# 6. COMPARACIÓN FOTOGRÁFICA A vs B
# =========================

fig, axes = plt.subplots(1, 2, figsize=(16, 9))

axes[0].imshow(imagen_a)
axes[0].set_title(
    "A — Encuadre cercano\n"
    "Protagonismo de la planta, cielo dorado como fondo",
    fontweight="bold"
)
axes[0].axis("off")

axes[1].imshow(imagen_b)
axes[1].set_title(
    "B — Encuadre abierto\n"
    "Contexto del balcón, más cielo y ciudad",
    fontweight="bold"
)
axes[1].axis("off")

plt.suptitle(
    "Comparación encuadres — mismo sujeto, misma luz",
    fontsize=13,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

## Respuestas a las preguntas guía

**¿La luz revela o esconde?**

Revela las texturas de las hojas de la sansevieria: sus patrones oscuros y los bordes amarillos se hacen visibles gracias a la iluminación rasante. Al mismo tiempo, esconde el interior oscuro de las macetas y el suelo del balcón, concentrando la atención en las plantas.

**¿Genera textura?**

Sí. La luz lateral-baja enfatiza las nervaduras y el relieve de las hojas, haciendo perceptible la superficie irregular de la sansevieria que con luz frontal quedaría completamente plana.

**¿Construye atmósfera?**

Sí. El cielo naranja degradado, los tonos cálidos y la suavidad de las sombras generan una atmósfera tranquila, íntima y contemplativa. La hora dorada añade una dimensión temporal a la imagen: la sensación de un momento fugaz del día.

**¿Cómo cambia el volumen?**

La luz baja y lateral crea diferencias de luminosidad entre las caras iluminadas (arriba-lateral) y las zonas en sombra (base y centro), otorgando tridimensionalidad a las hojas y a la maceta.

```python
# =========================
# 7. EXPORTACIÓN
# =========================

# Imagen original A
cv2.imwrite(
    "punto07_fotografia_luz_original.jpg",
    cv2.cvtColor(imagen_a, cv2.COLOR_RGB2BGR)
)

# Escala de grises
cv2.imwrite(
    "punto07_fotografia_luz_grises.jpg",
    imagen_a_gris
)

# Esquema de luz
cv2.imwrite(
    "punto07_esquema_luz.jpg",
    cv2.cvtColor(imagen_esquema, cv2.COLOR_RGB2BGR)
)

# Comparación final (A vs B)
fig, axes = plt.subplots(1, 2, figsize=(16, 9))
axes[0].imshow(imagen_a)
axes[0].set_title("Imagen original", fontweight="bold")
axes[0].axis("off")
axes[1].imshow(imagen_a_gris, cmap="gray")
axes[1].set_title("Escala de grises", fontweight="bold")
axes[1].axis("off")
plt.tight_layout()
plt.savefig(
    "punto07_comparacion.png",
    bbox_inches="tight",
    dpi=300
)
plt.close()

print("Exportación completada:")
print("  - punto07_fotografia_luz_original.jpg")
print("  - punto07_fotografia_luz_grises.jpg")
print("  - punto07_esquema_luz.jpg")
print("  - punto07_comparacion.png")
```
